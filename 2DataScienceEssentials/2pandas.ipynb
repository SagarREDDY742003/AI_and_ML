# PANDAS

### Pandas is a powerful open-source Python library used for data manipulation, analysis, and cleaning

## Key Features

### 1. Data Structures:
#####     Series: 1D labeled array (like a column in Excel).

```python
import pandas as pd
s=pd.Series([10,20,30], index=["a","b","c"])
print(s)
```

#####      DataFrame: 2D labeled data structure (like an Excel sheet or SQL table).

```python
data = {"Name":["Alice","Bob"], "Age":[25,30]}
df = pd.DataFrame(data)
print(df)
```

### 2. Data Handling:
##### Read/write from CSV, Excel, SQL, JSON, Dictionary etc.

```python
csv = pd.read_csv("../DataSet/student.csv")
print(csv.head(3))
csv.shape  # returns no of rows and columns
```

```python
exl = pd.read_excel("../DataSet/student.xlsx",nrows=3)
print(exl)
```

### 3.Data Operations:
#####      Viewing, Filtering, grouping, merging, reshaping.
#####      Statistical and mathematical operations.

```python
print(csv.head(3)) 
csv.describe()
```

```python
print(csv.tail(3))
```

```python
print(csv.info())
```

```python
print(csv.describe())
```

```python
print(csv[["name","mark"]].head())
```

```python
print(csv[csv["mark"]>88])
```

```python
print(csv.iloc[0]) # give first row by position
```

```python
print(csv.loc[0]) # give first row without index
```

```python
print(csv.iloc[:,2].head()) # give column by position
```

```python
print(csv.loc[:, "name"].head()) # give specific column without index
```
