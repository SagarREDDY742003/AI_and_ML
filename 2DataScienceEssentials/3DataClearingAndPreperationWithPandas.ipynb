# Handling Missing Values
### Methods to handle missing values
##### 1.Drop Missing values
Method: dropna()

Use case: When missing values are few and won’t affect the dataset significantly.

```python
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, None, 4],
    'B': [None, 2, 3, 4]
})
print("Original Data: \n",df,"\n")
```

```python
# Drop rows with any missing values
df_clean = df.dropna()
print("Data after Droping rows with missing values: \n",df_clean,"\n")
```

```python
# Drop columns with missing values
df_clean_cols = df.dropna(axis=1)
print("Data after Droping columns with missing values: \n",df_clean_cols)
```

##### 2.Fill Missing values
Method: fillna()

Use case: When you want to replace missing values with a specific value (like 0, mean, median, or forward/backward fill)

```python
# Fill with a constant 
df_filled = df.fillna(0) 
print(df_filled)
```

```python
# Fill with column mean
df['A'] = df['A'].fillna(df['A'].mean())
print(df) 
```

```python
# Forward fill (propagate last valid value forward) 
df_ffill = df.ffill()
print(df_ffill)
```

```python
# Backward fill 
df_bfill = df.bfill()
print(df_bfill)
```

##### 3.Interpolation
Method: interpolate()

Use case: When data is sequential (like time series) and you want to estimate missing values based on surrounding data

```python
# Linear interpolation
df_interp = df.interpolate(method='linear')
print(df_interp)
```

```python
import pandas as pd
import numpy as np

# Create a DataFrame with a datetime index
dates = pd.date_range("2024-01-01", periods=4, freq="D")
df = pd.DataFrame({
    "A": [1, np.nan, 3, 4],
    "B": [2, 3, np.nan, 5]
}, index=dates)

print("Original DataFrame:")
print(df)

# Time-based interpolation
df_time_interp = df.interpolate(method="time")
print("\nAfter time interpolation:")
print(df_time_interp)
```

### Data Transformations
##### 1.Renaming Columns
Method: rename() or directly assigning to df.columns

Use case: When column names are unclear, inconsistent, or need standardization.

```python
df = pd.DataFrame({
    "oldName1": [1, 2], 
    "oldName2": [3, 4]
})
print(df)
```

```python
# Rename specific columns 
df = df.rename(columns={'oldName1': 'NewName1', 'oldName2': 'NewName2'}) 
print(df)
```

```python
# Rename all columns at once 
df.columns = ['ColumnA', 'ColumnB']
print(df)
```

##### 2.Change Data Types
Method: astype()

Use case: Ensuring columns have the correct type (e.g., integers, floats, strings, datetime).

```python
df = pd.DataFrame({ 
    'A': ['1', '2', '3'], 
    'B': ['2024-01-01', '2024-01-02', '2024-01-03'] 
})
print(df.dtypes)
```

```python
# Convert string numbers to integers
df['A'] = df['A'].astype(int)
print(df.dtypes)
```

```python
# Convert to datetime
df['B'] = pd.to_datetime(df['B'])
print(df.dtypes)
```

##### 3.Creating or Modify Columns
Method: Direct assignment or using apply()

Use case: Deriving new features, applying transformations, or updating values.

```python
df = pd.DataFrame({ 'A': [10, 20, 30], 'B': [1, 2, 3] })
print(df)
```

```python
# Create a new column based on existing ones
df['C'] = df['A']+df['B']
print(df)
```

```python
# Modify an existing column
df['A'] = df['A']*2
print(df)
```

```python
# Create a column using a function
df['D'] = df['B'].apply(lambda x: x**2)
print(df)
```

### Combining and Merging DataFrames
##### 1.Concatenation
Method: pd.concat()

Use case: Stack DataFrames either vertically (rows) or horizontally (columns).

```python
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]}) 
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
```

```python
# Vertical concatenation (row-wise)
df_concat = pd.concat([df1,df2])
print(df_concat)
```

```python
# Vertical concatenation (row-wise)
df_concat_axis1 = pd.concat([df1,df2],axis=1)
print(df_concat_axis1)
```

##### 2.Merging
Method: pd.merge()

Use case: Combine DataFrames based on common columns or indices (like SQL joins).

```python
df1 = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['A', 'B', 'C']}) 
print("df1: \n",df1)
df2 = pd.DataFrame({'ID': [1, 2, 4], 'Score': [90, 80, 70]})
print("df2: \n",df2)
```

```python
# Inner join (only matching IDs)
df_inner = pd.merge(df1, df2, on='ID', how='inner')
print(df_inner)
```

```python
# Left join (all from df1, match from df2)
df_left = pd.merge(df1, df2, on='ID', how='left')
print(df_left)
```

```python
# Outer join (all IDs from both)
df_outer = pd.merge(df1, df2, on='ID', how='outer')
print(df_outer)
```

##### 3.Joining
Method: df.join()

Use case: Combine DataFrames based on their index (default) or a key column.

```python
df1 = pd.DataFrame({'Name': ['A', 'B', 'C']}, index=[1, 2, 3]) 
df2 = pd.DataFrame({'Score': [90, 80, 70]}, index=[1, 2, 4])
print("df1: \n",df1)
print("df2: \n",df2)
```

```python
# Default join on index
df_join = df1.join(df2, how='inner')
print(df_join)
```

```python
# Left join on index 
df_join_left = df1.join(df2, how='left')
print(df_join_left)
```
