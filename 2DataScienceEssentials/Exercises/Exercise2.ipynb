# 2. Select specific columns and filter rows

```python
import pandas as pd
# load data set
ds = pd.read_csv("../../DataSet/Iris.csv")
```

```python
selected_columns = ds[["Species","SepalLengthCm"]]
print(selected_columns.head())
```

```python
filter_rows = ds[(ds["SepalLengthCm"]>5.0) & (ds["Species"] == "Iris-setosa")]
print(filter_rows.head())
```

# 3. Save filtered data in new csv file.

```python
filter_rows.to_csv('../../DataSet/filtered_iris.csv',header=True,index=False)
df = pd.read_csv("../../DataSet/filtered_iris.csv")
print(df.head())
```

# 4. Create a DataFrame from a dictionary and add a new calculated column

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Math': [85, 92, 78, 88],
    'Science': [90, 85, 82, 95]
}
df = pd.DataFrame(data)
print(df)
```

```python
df['Average'] = (df['Math']+df['Science'])/2
print(df)
```
