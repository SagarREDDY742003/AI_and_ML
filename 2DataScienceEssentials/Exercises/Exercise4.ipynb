# Group data by a categorical column

```python
import pandas as pd

data = {
    "class":["A","B","A","B","C","C"],
    "score":[85,90,88,72,95,90],
    "Age":[15,16,15,17,16,15]
}

df = pd.DataFrame(data)
print(df)
```

```python
grouped = df.groupby("class").mean()
print(grouped)
```

# Calculate Summery Statestics for grouped data

```python
import pandas as pd

df = pd.DataFrame({
    "class":["A","B","A","B","C","C"],
    "score":[85,90,88,72,95,90],
    "age":[15,16,15,17,16,15]
})

stats = df.groupby("class").agg({
    "score":["mean","max","min"],
    "age":["mean","max","min"],
})
print(stats)
```

# Create a dataset of sales data and group it by region or product category

```python
import pandas as pd

# Sample sales data
data = {
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Product': ['Laptop', 'Laptop', 'Tablet', 'Tablet', 'Phone', 'Phone', 'Laptop', 'Phone'],
    'Sales': [1200, 1500, 800, 900, 2000, 1800, 1100, 2200]
}

df = pd.DataFrame(data)
print("Sales Data:")
print(df)
```

```python
# Total sales by region
sales_by_region = df.groupby('Region')['Sales'].sum()
print(sales_by_region)
```

```python
# Total sales by product
sales_by_product = df.groupby('Product')['Sales'].sum()
print(sales_by_product)
```

```python
# Group by both Region and Product
sales_summary = df.groupby(['Region','Product'])['Sales'].sum()
print(sales_summary)
```

# Use pivot_table to calculate total sales per region and per year

```python
import pandas as pd

# Sample sales data
data = {
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Year':   [2024, 2024, 2024, 2024, 2025, 2025, 2025, 2025],
    'Sales':  [1200, 1500, 800, 900, 2000, 1800, 1100, 2200]
}

df = pd.DataFrame(data)
print("Sales Data:")
print(df)
```

```python
# Pivot table: total sales per region and year
pivot = pd.pivot_table(
    df, 
    values='Sales', 
    index='Region', 
    columns='Year', 
    aggfunc='sum', 
    margins=True, 
    margins_name='Total'
)

print(pivot)
```

# create a custome aggregation fumction to calculate the variance for each group

```python
import pandas as pd

# Example dataset
df = pd.DataFrame({
    'Department': ['HR', 'IT', 'HR', 'Finance', 'IT', 'Finance'],
    'Salary': [50000, 60000, 52000, 70000, 65000, 72000]
})

print(df)
```

```python
def variance(x):
    return x.var()

dept_variance = df.groupby('Department')['Salary'].agg(variance)
print(dept_variance)
```

```python
dept_stats = df.groupby('Department')['Salary'].agg(['mean', 'min', 'max', variance])
print(dept_stats)
```
