# Clean a dataset by handling missing values and renaming columns

```python
import pandas as pd
import numpy as np

# create a sample dataset
data = {
    "Name":["Alice","Bob",np.nan,"David"],
    "Age":[25,np.nan,30,35],
    "Score":[85,90,np.nan ,88]
}
df = pd.DataFrame(data)
print(df)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Score'] = df['Score'].interpolate()
print(df)
```

```python
df = df.rename(columns={'Name':'Student_Name','Score':'Exam_Score'})
print(df)
```

# Merge two datasets and  perform data transformations

```python
df1 = pd.DataFrame({
    "ID":[1,2,3],
    "Name":["Alice","Bob","Charlie"],
    "Age":[25,30,35]
})
df2 = pd.DataFrame({
    "ID":[1,2,3],
    "Score":[85,90,88]
})
print("df1: \n",df1)
print("df2: \n",df2)
```

```python
merged = pd.merge(df1,df2,how="inner",on="ID")
print(merged)
```

```python
merged["Score_Percentage"] = (merged["Score"]/100)*100
print(merged)
```

# Drop Columns with moer than 50% missing values

```python
import pandas as pd 
import numpy as np 

df = pd.DataFrame({ 
    'A': [1, 2, np.nan, 4, np.nan], 
    'B': [np.nan, np.nan, np.nan, 4, 5], 
    'C': [1, 2, 3, 4, 5] 
})
print(df)
```

```python
# Drop columns with more than 50% missing values 
threshold = len(df) * 0.5 # 50% of total rows

df_clean = df.dropna(axis=1,thresh=threshold)
print("After dropping columns (>50% NaN): \n", df_clean)
```

# Merge 3 datasets and analyze relationships between them

```python
import pandas as pd

# Dataset 1: Students
students = pd.DataFrame({
    'StudentID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David']
})

# Dataset 2: Courses
courses = pd.DataFrame({
    'CourseID': [101, 102, 103],
    'CourseName': ['Math', 'Science', 'History']
})

# Dataset 3: Enrollments (relationship between students and courses)
enrollments = pd.DataFrame({
    'StudentID': [1, 2, 2, 3, 4],
    'CourseID': [101, 102, 103, 101, 102],
    'Grade': ['A', 'B', 'A', 'C', 'B']
})

print("students: \n",students)
print("\n courses: \n",courses)
print("\n enrollments: \n",enrollments)
```

```python
# Merge students with enrollments
student_courses = pd.merge(students, enrollments, on='StudentID', how='inner')
print(student_courses)
```

```python
# Merge with courses
final_df = pd.merge(student_courses, courses, on='CourseID', how='inner')
print(final_df)
```

```python
final_df.groupby('CourseName')['Grade'].value_counts()
```

```python
grade_map = {'A': 4, 'B': 3, 'C': 2, 'D': 1}
final_df['GradeValue'] = final_df['Grade'].map(grade_map)
print("After Grade mapping: \n",final_df)
avg_grades = final_df.groupby('Name')['GradeValue'].mean()
print("\n Average grades of each student: \n",avg_grades)
```

```python
course_popularity = final_df['CourseName'].value_counts()
print(course_popularity)
```

# Convert categorical data to numerical using one-hot encoding

One-Hot Encoding:

It transforms categorical values into binary columns (0/1).

Each category becomes a separate column.

A row gets a 1 in the column corresponding to its category, and 0 elsewhere.

```python
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'City': ['London', 'Paris', 'London', 'Berlin']
})

print("Original DataFrame:")
print(df)

# Apply one-hot encoding
df_encoded = pd.get_dummies(df, columns=['City'])

print("\nAfter One-Hot Encoding:")
print(df_encoded)
```
