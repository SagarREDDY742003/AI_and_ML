# Use Seaborn to create a violin plot or box plot for visulaization distributions

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Example dataset
tips = sns.load_dataset("tips")

# Violin plot
sns.violinplot(x="day", y="total_bill", data=tips, palette="muted")
plt.title("Distribution of Total Bill by Day (Violin Plot)")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Example dataset
tips = sns.load_dataset("tips")

# Box plot
sns.boxplot(x="day", y="total_bill", data=tips, palette="Set2")
plt.title("Distribution of Total Bill by Day (Box Plot)")
plt.show()
```

# Combine multiple plots in a single figure using matplotlib's subplot

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load example dataset
tips = sns.load_dataset("tips")

# Create a figure with 1 row and 2 columns of subplots
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Violin plot on the first subplot
sns.violinplot(x="day", y="total_bill", data=tips, palette="muted", ax=axes[0])
axes[0].set_title("Violin Plot: Total Bill by Day")

# Box plot on the second subplot
sns.boxplot(x="day", y="total_bill", data=tips, palette="Set2", ax=axes[1])
axes[1].set_title("Box Plot: Total Bill by Day")

# Adjust layout for better spacing
plt.tight_layout()
plt.show()

```
