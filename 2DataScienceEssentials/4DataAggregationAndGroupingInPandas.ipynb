# Grouping data by categories
Grouping data by categories in pandas is done using the powerful groupby() function. 
It allows you to split data into groups based on categorical values, apply aggregations, and analyze patterns.

### GroupBy in Pandas
Operations

    1.Iterate over groups
    
    2.Apply aggregation

```python
import pandas as pd 

df = pd.DataFrame({ 
    'Department': ['HR', 'IT', 'HR', 'Finance', 'IT', 'Finance'], 
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'], 
    'Salary': [50000, 60000, 52000, 70000, 65000, 72000] 
})

print(df)
```

```python
# Group by Department
grouped = df.groupby('Department')

# Iterate over groups
for dept, group in grouped:
    print(f"\nDepartment: {dept}")
    print(group)
```

```python
avg_salary = grouped['Salary'].mean()
print(avg_salary)

dept_stats = grouped['Salary'].agg(['mean','min','max'])
print(dept_stats)
```

# Aggregation Functions
### Using groupby
You can group data by categories and then apply aggregation functions

```python
import pandas as pd

df = pd.DataFrame({
    'Department': ['HR', 'IT', 'HR', 'Finance', 'IT', 'Finance'],
    'Salary': [50000, 60000, 52000, 70000, 65000, 72000]
})

# Group by Department and calculate mean salary
avg_salary = df.groupby('Department')['Salary'].mean()
print(avg_salary)

# Multiple aggregations
dept_stats1 = df.groupby('Department')['Salary'].agg(['mean', 'min', 'max'])
print(dept_stats1)

dept_stats2 = df.groupby('Department').agg({'Salary':['mean', 'min', 'max']})
print(dept_stats2)
```

### Using pivot_table
pivot_table() is similar to Excel pivot tables — great for summarizing data.

```python
# Pivot table: average salary per department
pivot = pd.pivot_table(df, values='Salary', index='Department', aggfunc='mean')
print(pivot)
```

```python
pivot_multi = pd.pivot_table(df, values='Salary', index='Department', aggfunc=['mean','max','min'])
print(pivot_multi)
```

### Custom Aggregation
##### Apply custom functions using agg()
You can apply your own functions to grouped data.

```python
# Custom aggregation: range of salaries (max - min)
def range_fun(x):
    return x.max()-x.min()

custom_agg = df.groupby('Department')['Salary'].agg(range_fun)
print(custom_agg)
```
