# Arbitary Keyword Arguments
# *kwargs

```python
def person_info(**info):
    print(type(info))
    for key, value in info.items():
        print(key,": ",value)

person_info(name="sagar",age=22,grade='A')
```

# use both *args and **Kwargs

```python
def person_info(*num,**info):
    for key, value in info.items():
        print(key,": ",value)
    print(num)

person_info(1,2,3,name="sagar",age=22,grade='A')
```

```python
def filter_and_square(numbers):
    """
    Return the squares of even numbers from the input list.
    
    Args:
        numbers (list): List of integers.
    
    Returns:
        list: Squares of even numbers.
    """
    # Lambda defines the transformation, list comprehension handles filtering
    square_even = lambda x: x ** 2
    return [square_even(num) for num in numbers if num % 2 == 0]

filter_and_square([1,2,3,4])
```

# Sum of squares of even numbers

```python
def sum_of_squares_of_evens(numbers):
    """
    Calculate the sum of squares of even numbers from a list.
    
    Args:
        numbers (list): List of integers.
    
    Returns:
        int: Sum of squares of even numbers.
    """
    square = lambda x: x ** 2
    return sum(square(num) for num in numbers if num % 2 == 0)

print(sum_of_squares_of_evens([1, 2, 3, 4, 5, 6]))
```

# Factorial using recursive functions

```python
def fact(n):
    if n==1 or n==0:
        return 1
    return n*fact(n-1)

fact(0)
```

# Binary Search

```python
def binary_search_recursion(num_list, num_to_find, left_index, right_index):\
    
    if right_index < left_index:
        return -1
    
    mid_index = (left_index + right_index) //2
    if mid_index >= len(num_list) or mid_index<0:
        return -1
    
    mid_num = num_list[mid_index]
    
    if mid_num == num_to_find:
        return mid_index
    
    if mid_num < num_to_find:
        left_index = mid_index+1
    else:
        right_index = mid_index-1
    
    return binary_search_recursion(num_list,num_to_find,left_index,right_index)

numbers = [1,2,3,4,5,6,7,8]
binary_search_recursion(numbers,1,0,len(numbers))
```

```python
def factorial(number):
    """
    Calculate the factorial of a non-negative integer using recursion.

    The factorial of a number n (denoted as n!) is defined as:
    - 0! = 1
    - n! = n × (n-1)! for n > 0
    
    Args:
        number (int): A non-negative integer whose factorial is to be computed.
    
    Returns:
        int: The factorial of the given number.
    
    Raises:
        ValueError: If the input is a negative integer.
    """
    if number < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    if number == 0:
        return 1
    return number * factorial(number - 1)
```

# Nested Functions and closure
## Closure: Inner function once returned carries the state of the parent function

```python
def traffic_light_validator():
    
    traffic_lights = ("RED", "GREEN", "ORANGE")
    
    def is_valid_light(color):
        if color.upper() in traffic_lights:
            return True
        else:
            return False
    return is_valid_light

color = "Red"
validator = traffic_light_validator()
if validator(color):
    print("YES")
else:
    print("NO")
```

```python
def multiply_by(factor):
    """
    Create a multiplier function using a closure.
    
    Args:
        factor (int or float): The number to multiply by.
     
    Returns:
        function: A lambda function that multiplies its input by the given factor. 
    """
    return lambda number : number * factor

multiplier_of_2 = multiply_by(2)
print(multiplier_of_2(5)) 
print(multiplier_of_2(10)) 
```

# First Class Functions

```python
# 1. Assigning a function to a variable
def greet(name):
    return f"Hello, {name}!"

say_hello = greet  # Assign function to a variable
print(say_hello("Alice"))  # Works like greet()

# 2. Passing a function as an argument
def call_function(func, value):
    return func(value)

print(call_function(greet, "Bob"))

# 3. Returning a function from another function
def make_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier  # Returning a function

times3 = make_multiplier(3)
print(times3(10))  # 30

# 4. Storing functions in a data structure
function_list = [greet, str.upper, str.lower]
for func in function_list:
    print(func("Python"))
```
