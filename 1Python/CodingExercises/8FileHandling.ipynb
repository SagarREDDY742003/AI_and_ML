```python
from pathlib import Path

def count_words_in_file(file_path: str) -> int:
    """
    Count the number of words in a text file.

    Args:
        file_path (str): Path to the file.

    Returns:
        int: Number of words in the file.
    """
    path = Path(file_path)

    if not path.exists():
        raise FileNotFoundError(f"File not found: {file_path}")

    try:
        # Use context manager to ensure file is closed automatically
        with path.open("r", encoding="utf-8") as file:
            content = file.read()
        return len(content.split())
    except Exception as e:
        raise RuntimeError(f"Error reading file {file_path}: {e}")

print(count_words_in_file("sample.txt"))
```

# provided Python code that reads a large text file, counts the frequency of each word, and returns the top 10 most frequent words.

```python
def top_words(file_path):
    word_count = {}
    with open(file_path, 'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word_count[word] =word_count.get(word,0)+1
    sorted_words = sorted(word_count.items(), key=lambda item: item[1], reverse=True)
    return sorted_words[:10]

print(top_words("sample.txt"))
```

```python
from collections import Counter
def count_words(filename):
    
    path = Path(filename)
    if not path.exists():
        raise FileNotFoundError(f"File not found: {filename}")
    
    try:
        with open(filename,'r') as file:
            word_counter = Counter(word for line in file for word in line.split())
        return dict(word_counter)
    except Exception as e:
        raise RuntimeError(f"Error reading file {filename}: {e}")
        

print(count_words("sample.txt"))
```

def process_large_file(file_path):

            result.append(byte * 2)

```python
from pathlib import Path
from typing import List

def process_large_file(file_path: str, chunk_size: int = 8192) -> List[int]:
    """
    Process a large binary file in chunks to avoid loading it fully into memory.
    Each byte is doubled (example transformation).

    Args:
        file_path (str): Path to the binary file.
        chunk_size (int): Number of bytes to read per iteration (default=8KB).

    Returns:
        List[int]: List of processed byte values.
    """
    path = Path(file_path)
    if not path.exists():
        raise FileNotFoundError(f"File not found: {file_path}")

    result = []
    try:
        with path.open("rb") as file:
            while chunk := file.read(chunk_size):
                # Use list comprehension for efficiency
                result.extend(byte * 2 for byte in chunk)
        return result
    except Exception as e:
        raise RuntimeError(f"Error processing file {file_path}: {e}")
```
