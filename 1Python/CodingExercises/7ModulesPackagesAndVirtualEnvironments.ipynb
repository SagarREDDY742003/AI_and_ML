```python
from datetime import datetime, timedelta
from typing import Union

def count_weekends(start_date_str: str, end_date_str: str) -> int:
    """
    Count the number of weekend days (Saturday and Sunday)
    between two given dates, inclusive.

    Args:
        start_date_str (str): Start date in 'YYYY-MM-DD' format.
        end_date_str (str): End date in 'YYYY-MM-DD' format.

    Returns:
        int: Number of weekend days between the two dates.
    """
    start_date = datetime.strptime(start_date_str, '%Y-%m-%d').date()
    end_date = datetime.strptime(end_date_str, '%Y-%m-%d').date()

    if start_date > end_date:
        raise ValueError("Start date must not be after end date.")

    # Total days in range
    total_days = (end_date - start_date).days + 1

    # Full weeks and remaining days
    full_weeks, extra_days = divmod(total_days, 7)

    # Each full week has exactly 2 weekend days
    weekend_count = full_weeks * 2

    # Handle leftover days beyond full weeks
    for i in range(extra_days):
        day = start_date + timedelta(days=full_weeks * 7 + i)
        if day.weekday() in (5, 6):  # Saturday=5, Sunday=6
            weekend_count += 1

    return weekend_count

print(count_weekends("2026-03-01", "2026-03-07"))  # Output: 2
print(count_weekends("2026-03-01", "2026-03-10"))  # Output: 4
print(count_weekends("2026-03-06", "2026-03-06"))  # Output: 1 (Friday → 0, Saturday → 1)
```

```python
import subprocess
import sys
from pathlib import Path

def install_requirements(file_path: str) -> None:
    """
    Install packages listed in a requirements.txt file.

    Args:
        file_path (str): Path to the requirements file.
    """
    req_file = Path(file_path)

    if not req_file.exists():
        raise FileNotFoundError(f"Requirements file not found: {file_path}")

    try:
        # Install all packages in one pip call for efficiency
        subprocess.run(
            [sys.executable, "-m", "pip", "install", "-r", str(req_file)],
            check=True
        )
        print("✅ All packages installed successfully.")
    except subprocess.CalledProcessError as e:
        print(f"❌ Error occurred while installing packages: {e}")
    except Exception as e:
        print(f"⚠️ Unexpected error: {e}")

# Example usage
if __name__ == "__main__":
    install_requirements("requirements.txt")
```
