# Design an algorithm in Python that takes a list of mixed data types (integers, floats, and strings) and returns a dictionary categorizing each element by its data type. The algorithm should handle nested lists and ensure all elements are categorized correctly.

```python
def categorize_elements(data):
    """
    Categorize elements in a list by their data type.
    Handles nested lists recursively.
    
    Args:
        data (list): A list containing integers, floats, strings, or nested lists.
    
    Returns:
        dict: A dictionary with keys 'int', 'float', 'str' and lists of elements.
    """
    
    categories = {"int":[], "float":[], "string":[]}
    
    def process_element(element):
        # If the element is a list, process its contents recursively
        if isinstance(element, list):
            for item in element:
                process_element(item)
        elif isinstance(element,int):
            categories["int"].append(element)
        elif isinstance(element,float):
            categories["float"].append(element)
        elif isinstance(element,str):
            categories["string"].append(element)
    for element in data:
        process_element(element)
        
    return categories

data = [1, 2.5, "hello", [3, "world", [4.2, "nested", 7]], "end"]
result = categorize_elements(data)
print(result)
```

# Design an algorithm in Python that takes a list of complex numbers and returns a new list containing only those numbers whose magnitude is greater than a given threshold. Implement this algorithm and test it with the list [(1+1j), (3+4j), (0+2j)] and a threshold of 2.5.

```python
def filter_complex_basedOn_magnitude(numbers, threshold):
    """ 
    Return a list of complex numbers whose magnitude is greater than the threshold. 
    Args: 
        numbers (list of complex): List of complex numbers. 
        threshold (float): Minimum magnitude required. 
    Returns: 
        list of complex: Filtered list of complex numbers. 
    """ 
    # Use list comprehension with abs() to compute magnitude 
    # return [num for num in numbers if abs(num) > threshold]
    result=[]
    for number in numbers:
        if abs(number) > threshold:
            result.append(number)
        
    return result

numbers = [(1+1j), (3+4j), (0+2j)]
threshold = 2.5
result = filter_complex_basedOn_magnitude(numbers, threshold)
print(result)
```

```python
def process_string(s):
    """
    Process a string, ensuring escape characters are preserved
    and raw strings are handled correctly.
    """
    # Use join instead of repeated concatenation for efficiency
    return "".join(s)

# Example with escape characters
input_string = "Hello\nWorld"
print(process_string(input_string))

# Example with raw string (backslashes preserved literally)
raw_input = r"Hello\nWorld"
print(process_string(raw_input))
```

# Python code to processes a multiline string to count the frequency of each word across all lines.

```python
def word_frequency(multiline_str):
    """
    Count the frequency of each word in a multiline string.
    Words are normalized to lowercase.
    """
    word_count = {}
    
    # Split on whitespace (handles spaces, tabs, and newlines)
    for word in multiline_str.split():
        word = word.lower()
        
        word_count[word] = word_count.get(word,0) + 1    
    
    return word_count


text = """Hello world
Hello Python
world of code"""
print(word_frequency(text))
```
