# give the names and grades for each student in a class of n students, store them in nested lost and print the names of any students having the second lowest grade

```python
def sort_students(names_scores):
    scores = []
    for student in names_scores:
        scores.append(student[1])
    sorted_scores = sorted(set(scores))
    second_lowest = sorted_scores[1]
    
    names_second_lowest = []
    for student in names_scores:
        if student[1] == second_lowest:
            names_second_lowest.append(student[0])
    names_second_lowest = sorted(names_second_lowest)
    
    for name in names_second_lowest:
        print(name)


if __name__ == '__main__':
    names_scores = []
    for _ in range(int(input())):
        name = input()
        score = float(input())
        names_scores.append([name,score])
    sort_students(names_scores)
```

```python
def flatten_nested_list(nested_list):
    """
    Flatten a nested list into a single list while preserving order.
    
    Args:
        nested_list (list): A list that may contain other lists.
    
    Returns:
        list: A flattened list with all elements in order.
    """
    flat_list = []
    
    def _flatten(sublist):
        for item in sublist:
            if isinstance(item, list):
                _flatten(item)  # Recursively flatten nested lists
            else:
                flat_list.append(item)
    
    _flatten(nested_list)
    return flat_list


nested_list  = [[1,2,3],[4,5],[6]]
result = flatten_nested_list(nested_list)
print(result)
```

# Design an algorithm to sort a list of tuples, where each tuple contains a name and a score, first by score in descending order and then by name in ascending order if scores are equal.

### data = [('Alice', 88), ('Bob', 75), ('Charlie', 88), ('David', 92)]

```python
data = [('Alice', 88), ('Bob', 75), ('Charlie', 88), ('David', 92)]

# Sort by score descending, then name ascending
sorted_data = sorted(data, key=lambda x: (-x[1], x[0]))

print(sorted_data)
```

# Find the duplicates

```python
def find_duplicates(lst):
    """
    Find duplicate elements in a list.
    
    Args:
        lst (list): Input list.
    
    Returns:
        list: A list of duplicate elements (unique, order preserved).
    """
    seen = {}
    duplicates = []
    
    for char in lst:
        # Increment count using dict.get()
        seen[char] = seen.get(char, 0) + 1
        
        # Add to duplicates only once when count reaches 2
        if seen[char] == 2:
            duplicates.append(char)
    
    return duplicates

data = ['a', 'b', 'c', 'a', 'b', 'd', 'a']
print(find_duplicates(data))
```
