````python
def set_operations(set1, set2):
    union = set1 | set2
    intersection = set1 & set2
    difference = set1 - set2
    
    return union,intersection,difference

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

result = set_operations(set_a, set_b)
print(result)
print("Union:", result[0])
print("Intersection:", result[1])
print("Difference:", result[2])
````

Here’s a clear algorithm in Python that computes the **union**, **intersection**, and **difference** of two sets of integers, along with an explanation of the approach and its time complexity:

```python
def set_operations(set1, set2):
    """
    Perform set operations on two sets of integers.
    
    Args:
        set1 (set): First set of integers.
        set2 (set): Second set of integers.
    
    Returns:
        tuple: (union, intersection, difference)
            - union: elements in either set1 or set2
            - intersection: elements common to both sets
            - difference: elements in set1 but not in set2
    """
    union = set1 | set2          # Union using | operator
    intersection = set1 & set2   # Intersection using & operator
    difference = set1 - set2     # Difference using - operator
    
    return union, intersection, difference
```

---

### 🔑 Approach
- **Union (`|`)**: Combines all unique elements from both sets.  
- **Intersection (`&`)**: Keeps only elements present in both sets.  
- **Difference (`-`)**: Keeps elements in `set1` that are not in `set2`.  
- Python’s built-in set operators are optimized and expressive, making the code concise and efficient.

---

### ⏱️ Time Complexity
Let \(n = |set1|\) and \(m = |set2|\):

- **Union**: \(O(n + m)\) — each element from both sets is processed once.  
- **Intersection**: \(O(\min(n, m))\) — checks membership across sets.  
- **Difference**: \(O(n)\) — each element in `set1` is checked against `set2`.  

Overall complexity: **\(O(n + m)\)**.  
Space complexity: also **\(O(n + m)\)**, since new sets are created for results.

---

### ✅ Example Run
```python

set_b = {3, 4, 5, 6}

result = set_operations(set_a, set_b)
print("Union:", result[0])
print("Intersection:", result[1])
print("Difference:", result[2])
```

# Design an algorithm in Python that takes a list of tuples, where each tuple contains a student's name and their score. The algorithm should return a set of names of students who scored above a given threshold. Consider efficiency and edge cases in your design.

````python
def students_above_threshold(data, threshold):
    """
    Return a set of student names who scored above a given threshold.
    
    Args:
        data (list of tuples): Each tuple contains (name, score).
        threshold (int): The score threshold.
    
    Returns:
        set: Names of students with scores above the threshold.
    """
    # Use set comprehension for efficiency and readability
    return {name for name, score in data if score > threshold}


data = [('Alice', 88), ('Bob', 75), ('Charlie', 92), ('David', 60)]
threshold = 80

print(students_above_threshold(data, threshold))
````

# Design an algorithm in Python that takes a list of tuples, where each tuple contains a student's name and their scores in three subjects. The algorithm should return a set of names of students who have an average score above 75. Use tuple unpacking and set operations in your solution.

````python
def high_achievers(students_scores):
    """
    Return a set of student names whose average score
    across three subjects is above 75.
    
    Args:
        students_scores (list of tuples): Each tuple contains
            (name, score1, score2, score3).
    
    Returns:
        set: Names of students with average score > 75.
    """
    # Use tuple unpacking and set comprehension
    return {name for name, s1, s2, s3 in students_scores if (s1 + s2 + s3) / 3 > 75}

students = [
    ('Alice', 80, 90, 85),
    ('Bob', 70, 65, 72),
    ('Charlie', 88, 76, 92),
    ('David', 60, 70, 68)
]

print(high_achievers(students))
````

# Refactor the given code to efficiently remove duplicates from a list while maintaining the original order of elements.

````python
def remove_duplicates(lst):
    """
    Remove duplicates from a list while maintaining order.
    
    Args:
        lst (list): Input list with possible duplicates.
    
    Returns:
        list: A new list with duplicates removed, preserving order.
    """
    seen = set()
    return [x for x in lst if not (x in seen or seen.add(x))]

data = [3, 1, 2, 3, 4, 2, 5, 1]
print(remove_duplicates(data))
````

# Design an algorithm in Python that takes two sets and returns a new set containing elements that are present in either of the sets but not in both (symmetric difference). Optimize the algorithm for performance with large datasets.

````python
def symmetric_difference(set1, set2):
    """
    Return the symmetric difference of two sets:
    elements present in either set1 or set2 but not in both.
    
    Args:
        set1 (set): First set of integers.
        set2 (set): Second set of integers.
    
    Returns:
        set: Symmetric difference of set1 and set2.
    """
    # Use the built-in symmetric difference operator ^
    return set1 ^ set2

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

result = symmetric_difference(set_a, set_b)
print(result)
````

# Design an algorithm in Python that takes a list of tuples, where each tuple contains two sets, and returns a list of tuples with the union, intersection, and difference of each pair of sets. Implement this algorithm.

````python
def process_tuple_sets(tuple_list):
    """
    Given a list of tuples, where each tuple contains two sets,
    return a list of tuples with the union, intersection, and difference
    of each pair of sets.

    Args:
        tuple_list (list of tuples): Each tuple is (set1, set2).

    Returns:
        list of tuples: Each tuple is (union, intersection, difference).
    """
    return [(s1|s2, s1&s2,s1-s2) for s1,s2 in tuple_list]

data = [
    ({1, 2, 3}, {3, 4, 5}),
    ({10, 20}, {20, 30}),
    ({7, 8}, {9, 10})
]

result = process_tuple_sets(data)
for idx, (union, intersection, difference) in enumerate(result, 1):
    print(f"Pair {idx}:")
    print("  Union:", union)
    print("  Intersection:", intersection)
    print("  Difference:", difference)
````

# Design an algorithm that efficiently checks if two lists contain the same unique elements, using frozen sets to ensure immutability and optimize performance.

````python
def check_unique_elements(list1, list2):
    """
    Check if two lists contain the same unique elements.
    
    Args:
        list1 (list): First list of integers (or hashable elements).
        list2 (list): Second list of integers (or hashable elements).
    
    Returns:
        bool: True if both lists contain the same unique elements, False otherwise.
    """
    # Convert lists to frozensets (immutable sets of unique elements)
    return frozenset(list1) == frozenset(list2)

list_a = [1, 2, 3, 3, 4]
list_b = [4, 3, 2, 1]

print(check_unique_elements(list_a, list_b))  # True

list_c = [1, 2, 3]
list_d = [1, 2, 4]

print(check_unique_elements(list_c, list_d))  # False
````

# Design an algorithm using Python sets to find the intersection of multiple sets and explain how this operation can be optimized for large datasets.

````python
def find_intersection(*sets):
    """
    Find the intersection of multiple sets.
    
    Args:
        *sets (set): Variable number of sets.
    
    Returns:
        set: Intersection of all input sets.
    """
    if not sets:
        return set()
    
    sorted_sets = sorted(sets,key=len)
    intersection_result = sorted_sets[0].copy()
    
    for s in sorted_sets:
        intersection_result &=s
    
    return intersection_result

set_a = {1, 2, 3, 4}
set_b = {2, 3, 4, 5}
set_c = {3, 4, 6}

print(find_intersection(set_a, set_b, set_c))
````
