### Design a Python class that demonstrates encapsulation by managing a collection of items with controlled access. Include methods to add, remove, and list items, ensuring the collection cannot be modified directly.
#### Design a Python class named ItemCollection that encapsulates a list of items. Implement methods to add an item, remove an item, and list all items. Ensure the list of items cannot be accessed or modified directly from outside the class.

````python
class ItemCollection:
    """
    A class that encapsulates a collection of items.
    Provides controlled access via add, remove, and list methods.
    """

    def __init__(self):
        # Private attribute: cannot be accessed directly outside the class
        self.__items = []

    def add_item(self, item):
        """Add an item to the collection."""
        self.__items.append(item)

    def remove_item(self, item):
        """Remove an item from the collection. Returns True if removed, False if not found."""
        if item in self.__items:
            self.__items.remove(item)
            return True
        return False

    def list_items(self):
        """Return a copy of the items list to prevent direct modification."""
        return self.__items.copy()


collection = ItemCollection()

collection.add_item("Book")
collection.add_item("Pen")
collection.add_item("Pencil")

print(collection.list_items())

collection.remove_item("Pen")
print(collection.list_items())
````

#### Design a class hierarchy for a library management system that includes classes for 'Book', 'Member', and 'Loan'. Use inheritance to show relationships between these classes and implement the __init__ method appropriately for each class.

````python
from datetime import date

class LibraryEntity:
    """
    Base class for all entities in the library system.
    Provides a common identifier and display method.
    """
    def __init__(self, entity_id: int):
        self.entity_id = entity_id

    def display_info(self) -> str:
        return f"Entity ID: {self.entity_id}"


class Book(LibraryEntity):
    """
    Represents a book in the library.
    """
    def __init__(self, entity_id: int, title: str, author: str, year: int, isbn: str):
        super().__init__(entity_id)
        self.title = title
        self.author = author
        self.year = year
        self.isbn = isbn
        self.is_available = True

    def display_info(self) -> str:
        return f"Book[{self.entity_id}] - {self.title} by {self.author} ({self.year}), ISBN: {self.isbn}"


class Member(LibraryEntity):
    """
    Represents a library member.
    """
    def __init__(self, entity_id: int, name: str, membership_date: date):
        super().__init__(entity_id)
        self.name = name
        self.membership_date = membership_date

    def display_info(self) -> str:
        return f"Member[{self.entity_id}] - {self.name}, Joined: {self.membership_date}"


class Loan(LibraryEntity):
    """
    Represents a loan transaction between a member and a book.
    """
    def __init__(self, entity_id: int, book: Book, member: Member, loan_date: date, due_date: date):
        super().__init__(entity_id)
        self.book = book
        self.member = member
        self.loan_date = loan_date
        self.due_date = due_date
        self.returned = False

        # Mark book as unavailable when loan is created
        if not book.is_available:
            raise Exception(f"Book '{book.title}' is already loaned out.")
        book.is_available = False

    def mark_returned(self) -> None:
        """Mark the loan as returned and make the book available again."""
        self.returned = True
        self.book.is_available = True

    def display_info(self) -> str:
        status = "Returned" if self.returned else "On Loan"
        return (f"Loan[{self.entity_id}] - {self.book.title} loaned to {self.member.name} "
                f"on {self.loan_date}, due {self.due_date}, Status: {status}")


# Create book and member
book1 = Book(1, "1984", "George Orwell", 1949, "9780451524935")
member1 = Member(101, "Alice", date(2025, 5, 20))

# Loan the book
loan1 = Loan(1001, book1, member1, date(2026, 3, 10), date(2026, 3, 24))

print(book1.display_info())
print(member1.display_info())
print(loan1.display_info())

# Return the book
loan1.mark_returned()
print(loan1.display_info())
````

#### Design a Python-based system for managing a zoo, including classes for Animal, Mammal, Bird, and Reptile. Implement method overriding to handle feeding behavior specific to each type of animal.

````python
class Animal:
    """
    Base class for all animals in the zoo.
    """
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def feed(self):
        """Generic feeding behavior (to be overridden)."""
        return f"{self.name} the {self.species} is being fed."


class Mammal(Animal):
    """
    Mammal subclass with specific feeding behavior.
    """
    def __init__(self, name, species, diet = "milk/vegetation"):
        super().__init__(name, species)
        self.diet = diet

    def feed(self):
        return f"{self.name} the {self.species} (Mammal) is fed with {self.diet}."


class Bird(Animal):
    """
    Bird subclass with specific feeding behavior.
    """
    def __init__(self, name, species, diet = "seeds/insects"):
        super().__init__(name, species)
        self.diet = diet

    def feed(self):
        return f"{self.name} the {self.species} (Bird) pecks at {self.diet}."


class Reptile(Animal):
    """
    Reptile subclass with specific feeding behavior.
    """
    def __init__(self, name, species, diet = "insects/meat"):
        super().__init__(name, species)
        self.diet = diet

    def feed(self):
        return f"{self.name} the {self.species} (Reptile) consumes {self.diet}."

lion = Mammal("Leo", "Lion", "meat")
parrot = Bird("Polly", "Parrot", "fruits and seeds")
snake = Reptile("Slither", "Python")

animals = [lion, parrot, snake]

for animal in animals:
    print(animal.feed())
````

#### Design a class hierarchy for a university system. The system should include: - A `Person` class with basic attributes like `name` and `age`. - A `Student` class that inherits from `Person` and includes attributes like `student_id` and `courses`. - A `Professor` class that inherits from `Person` and includes attributes like `employee_id` and `courses_taught`. - A `Course` class with attributes like `course_name`, `course_code`, and `participants`. - Demonstrate polymorphism by having a method `get_details` that behaves differently in `Student` and `Professor` classes.

````python
class Person:
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def get_details(self):
        return f"Person: {self.name}, Age:{self.age}"
    
class Course:
    
    def __init__(self, course_name, course_code):
        self.course_name = course_name
        self.course_code = course_code
        self.participants = [] # list of student objects
        self.professors = [] # list of professor objects
    
    def add_participants(self, student):
        self.participants.append(student)
        
    def add_professors(self, professor):
        self.professors.append(professor)
    
    def get_details(self):
        participant_names = [s.name for s in self.participants]
        professor_names = [s.name for s in self.professors]
        return f"Course: {self.course_name} ({self.course_code}), Participants: {participant_names}, Professors: {professor_names}"
    
class Student(Person):
    
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id
        self.courses = [] # list of course objects
        
    def enroll(self, course):
        self.courses.append(course)
        course.add_participants(self)
    
    def get_details(self):
        course_names = [c.course_name for c in self.courses]
        return f"Student: {self.name}, ID: {self.student_id}, Courses: {course_names}"
    
class Professor(Person):
    
    def __init__(self, name, age, employee_id):
        super().__init__(name, age)
        self.employee_id = employee_id
        self.courses_taught = [] # list of course objects
        
    def assign_course(self, course):
        course.add_professors(self)
        self.courses_taught.append(course)
    
    def get_details(self):
        course_names = [c.course_name for c in self.courses_taught]
        return f"Professor: {self.name}, Employee ID: {self.employee_id}, Courses Taught: {course_names}"
    
# Create courses
course1 = Course("Computer Science", "CS101")
course2 = Course("Mathematics", "MATH201")

# Create student and professor
student1 = Student("Alice", 20, "S123")
professor1 = Professor("Dr. Smith", 45, "P456")

# Enroll student and assign professor
student1.enroll(course1)
professor1.assign_course(course1)
professor1.assign_course(course2)

# Display details
print(student1.get_details())
print(professor1.get_details())
print(course1.get_details())
print(course2.get_details())
````

#### Design a system using Python that demonstrates polymorphism through a hierarchy of classes representing different types of bank accounts. Describe your design and how polymorphism is achieved.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            raise ValueError("Insufficient amount")
        
    def get_balance(self):
        return self.balance
    
    def __str__(self):
        return f"{self.owner}'s account balance: {self.balance}"

class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, interest_rate=0.01):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def apply_interest(self):
        self.balance += self.balance * self.interest_rate
        
    # Polymorphic override
    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Cannot withdraw more than balance in SavingsAccount")
        super().withdraw(amount)
            

class CheckingAccount(BankAccount):
    def __init__(self, owner, balance=0, overdraft_limit=100):
        super().__init__(owner, balance)
        self.overdraft_limit = overdraft_limit

    # Polymorphic override
    def withdraw(self, amount):
        if self.balance - amount >= -self.overdraft_limit:
            self.balance -= amount
        else:
            raise ValueError("Overdraft limit exceeded")
        
accounts = [
    SavingsAccount("Alice", 500, 0.05),
    CheckingAccount("Bob", 200, 150),
]

for account in accounts:
    account.withdraw(100)   # Same method name, different behavior!
    print(account)
````
