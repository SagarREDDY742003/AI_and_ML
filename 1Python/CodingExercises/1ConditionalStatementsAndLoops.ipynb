# Design an algorithm in Python that uses nested loops and conditional statements to generate a multiplication table for numbers 1 through n, but only includes products that are even numbers.

```python
def even_multiplication_table(n):
    
    '''
    Generate a multiplication table for numbers 1 througn n,
    but only include products that are even numbers.
    
    Args: 
        n(int): The upper limit of multiplication table.
        
    Returns:
        list of list: A nested list representing the table of even products.
    '''
    
    table = []  # Outer list to hold rows of the table
    
    # Outer loop: iterate through rows (multiplicands)
    for i in range(1, n + 1):
        row = []  # Inner list to hold products for this row
        
        # Inner loop: iterate through columns (multipliers)
        for j in range(1, n + 1):
            product = i * j
            
            # Conditional check: include only even products
            if product % 2 == 0:
                row.append(product)
        
        # Add the completed row to the table
        table.append(row)
        
    return table

# Example usage:
n = int(input("Enter a number: "))
result = even_multiplication_table(n)

# Print the table in a readable format
for row in result:
    print(row)
```

# Design an algorithm using loops and conditional statements to find the longest sequence of consecutive positive numbers in a list. The algorithm should return the length of this sequence.

```python
def longest_positive_sequence(numbers):
    
    '''
    Find the longest sequence of consecutive positive numbers in a list.
    
    Args:
        numbers (list of int): Input list of integers.
    
    Returns:
        int: Length of the longest consecutive positive sequence. 
    '''
    
    max_len = 0  # Tracks the longest sequence found
    current_len = 0  # Tracks the current sequence length
    
    # Iterate through each number in the list
    for num in numbers:
        if num > 0:
            # If the number is positive, extend the current sequence
            current_len += 1
            # Update max_length if current sequence is longer
            max_len = max(current_len,max_len)
        else:
            # Reset current sequence length when a non-positive number is encountered
            current_len = 0
    
    return max_len

numbers = [1, 2, -3, 4, 5, 6, -1, 2, 3]
print(longest_positive_sequence(numbers))
# Output: 3 (sequence [4, 5, 6])
```

#  The code calculates the sum of squares of even numbers in a list.

```python
def sum_of_squares(numbers):
    """
    Calculate the sum of squares of even numbers in a list.
    
    Args:
        numbers (list of int): Input list of integers.
    
    Returns:
        int: Sum of squares of even numbers.
    """
    
    '''
    sum = 0
    for num in numbers:
        if num % 2 == 0:
            sum += num**2
    return sum
    '''
    
    # Use a generator expression with sum() for concise and efficient calculation
    return sum(num ** 2 for num in numbers if num % 2 == 0)

numbers = [1,2,3,4,5,6]
print(sum_of_squares(numbers))
```

# Design an algorithm using a while loop to simulate a simple banking system where a user can deposit, withdraw, and check balance. The system should continue to prompt the user for actions until they choose to exit. The system should start with a balance of 0. Users can input 'deposit' to add money, 'withdraw' to remove money, 'balance' to check the current balance, and 'exit' to end the session. Ensure that withdrawals do not exceed the current balance.

```python
def banking_system():
    """
    Simulate a simple banking system where users can:
    - deposit money
    - withdraw money
    - check balance
    - exit the system
    
    The system ensures withdrawals do not exceed the current balance.
    """
    balance = 0  # Initial balance
    
    while True:
        # Prompt user for an action
        action = input("Enter action (deposit, withdraw, balance, exit): ").strip().lower()
        
        if action == "deposit":
            # Ask for deposit amount
            amount = float(input("Enter amount to deposit: "))
            balance += amount
            print(f"Deposited {amount:.2f}. New balance: {balance:.2f}")
        
        elif action == "withdraw":
            # Ask for withdrawal amount
            amount = float(input("Enter amount to withdraw: "))
            if amount <= balance:
                balance -= amount
                print(f"Withdrew {amount:.2f}. New balance: {balance:.2f}")
            else:
                print("Insufficient funds. Withdrawal denied.")
        
        elif action == "balance":
            # Display current balance
            print(f"Current balance: {balance:.2f}")
        
        elif action == "exit":
            # End the session
            print("Exiting banking system. Goodbye!")
            break
        
        else:
            # Handle invalid input
            print("Invalid action. Please enter deposit, withdraw, balance, or exit.")
        

banking_system()   
```

# Design an algorithm in Python that processes a list of integers. The algorithm should sum the numbers until a negative number is encountered, at which point it should break out of the loop. If a zero is encountered, it should skip that number and continue. Return the sum of the processed numbers.

```python
def sum_until_negative(numbers):
    """
    Sum numbers in a list until a negative number is encountered.
    - Stop summing when a negative number appears.
    - Skip zeros (do not add them).
    
    Args:
        numbers (list of int): Input list of integers.
    
    Returns:
        int: Sum of processed numbers.
    """
    total = 0
    index = 0
    
    # Use a while loop to iterate through the list
    while index < len(numbers):
        num = numbers[index]
        
        if num < 0:
            # Break out of the loop when a negative number is found
            break
        elif num == 0:
            # Skip zero and continue to the next number
            index += 1
            continue
        else:
            # Add positive numbers to the total
            total += num
        
        index += 1
    
    return total


numbers = [5, 10, 0, 3, -2, 7]
print(sum_until_negative(numbers))
# Output: 18 (5 + 10 + 3, stops at -2)
```

```python
n=5
for i in range(1, n + 1):
    # Print leading spaces
    for j in range(n - i):
        print(" ", end="")
    
    # Print asterisks for the current row
    for k in range(1, 2*i):
        print("*", end="")
    print()
```

```python
# Function to print a half pyramid pattern
def half_pyramid(n):
    for i in range(1, n + 1):
        for j in range(1, i + 1):
            print("* ", end="")
        print("")

# Example: Print a half pyramid with 5 rows
n = 5
half_pyramid(n)
```

```python
# Function to print inverted half pyramid pattern
def inverted_half_pyramid(n):
    for i in range(n, 0, -1):
        for j in range(1, i + 1):
            print("* ", end="")
        print("\r")

# Example: Inverted Half Pyramid with n = 5
n = 5
inverted_half_pyramid(n)
```

```python
def hollow_pyramid(n):
    for i in range(1, n + 1):
        for j in range(1, 2 * n):
            if j == n - i + 1 or j == n + i - 1 or i == n:
                print("*", end="")
            else:
                print(" ", end="")
        print()

# Set the number of rows for the pyramid
rows = 5

# Call the function with the specified number of rows
hollow_pyramid(rows)
```

```python
def print_hollow_inverted_half_pyramid(rows):
    for i in range(rows, 0, -1):
        for j in range(rows - i):
            print(" ", end="")
        for j in range(i):
            if j == 0 or j == i - 1 or i == rows:
                print("*", end="")
            else:
                print(" ", end="")
        print()

# Example usage
num_rows = 5
print("Hollow Inverted Half Pyramid:")
print_hollow_inverted_half_pyramid(num_rows)
```

```python
# Function to print inverted full pyramid pattern
def inverted_full_pyramid(n):
    # Outer loop for the number of rows
    for i in range(n, 0, -1):
        # Inner loop for leading spaces in each row
        for j in range(n - i):
            print(" ", end="")
        # Inner loop for printing asterisks in each row
        for k in range(2*i - 1):
            print("*", end="")
        # Move to the next line after each row
        print("")

# Set the value of n (number of rows)
n = 5

# Call the function to print the inverted full pyramid
inverted_full_pyramid(n)
```

```python
def print_number_pyramid(rows):
    for i in range(1, rows + 1):
        # Print spaces
        for j in range(rows - i):
            print(" ", end="")
        # Print numbers
        for j in range(2 * i - 1):
            print(j + 1, end="")
        # Move to the next line after each row
        print()

# Example usage
num_rows = 5
print_number_pyramid(num_rows)
```

# Design an algorithm in Python that uses nested loops to find all pairs of numbers in a list that sum up to a given target value. Optimize the algorithm to avoid unnecessary comparisons.

```python
def find_pairs(numbers, target): 
    """ 
    Find all pairs of numbers in a list that sum up to a given target. 
    
    Args: 
        numbers (list of int): Input list of integers. 
        target (int): Target sum value. 
    
    Returns: 
        list of tuple: List of pairs (a, b) that sum to target. 
    """
    
    pairs = []
    
    # Outer loop: iterate through each element
    for i in range(len(numbers)):
        # Inner loop: start from the next element to avoid duplicate comparisons
        for j in range(i+1,len(numbers)):
            if numbers[i] + numbers[j] == target:
                pairs.append((numbers[i],numbers[j]))
    
    return pairs

numbers = [2, 4, 3, 5, 7, 8]
target = 10
print(find_pairs(numbers, target))
```

# Design an algorithm using a 'for' loop with 'else' to find the first non-repeating character in a string. If all characters repeat, return 'None'. Provide a detailed explanation of your approach.
## Example input: 'swiss' Expected output: 'w'

```python
def first_non_repeating_char(s):
    """
    Finds the first non-repeating character in a string using a for-else loop.
    Returns the character if found, otherwise returns 'None'.
    """

    # Outer loop: iterate through each character
    for i in range(len(s)):
        # Inner loop: check if this character repeats elsewhere
        for j in range(len(s)):
            if i != j and s[i] == s[j]:
                break  # Found a duplicate, stop checking this character
        else:
            # This else belongs to the inner for-loop
            # It executes only if the inner loop did NOT break
            return s[i]

    # If no non-repeating character found
    return None

print(first_non_repeating_char("swiss")) 
print(first_non_repeating_char("aabb"))
print(first_non_repeating_char(""))
```

# Write a Python function that uses logical operators to filter a list of numbers, returning only those that are even and greater than a specified threshold. Provide a test case to demonstrate your function.

```python
def filter_even_above_threshold(numbers, threshold):
    """
    Filter a list of numbers, returning only those that are:
    - Even
    - Greater than the specified threshold
    
    Args:
        numbers (list of int): Input list of integers.
        threshold (int): The minimum value a number must exceed.
    
    Returns:
        list of int: Filtered list of numbers.
    """
    # Use list comprehension with logical operators
    return [num for num in numbers if num%2==0 and num > threshold]

numbers = [5, 12, 18, 7, 22, 30, 3, 100]
threshold = 20

result = filter_even_above_threshold(numbers, threshold)
print(result)
```
