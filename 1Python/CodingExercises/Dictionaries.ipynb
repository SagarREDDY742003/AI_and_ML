```python
def char_frequency(s):
    """
    Calculate the frequency of each character in a string.
    
    Args:
        s (str): Input string.
    
    Returns:
        dict: A dictionary mapping characters to their frequency.
    """
    freq = {}
    for char in s:
        # Use dict.get() to simplify increment logic
        freq[char] = freq.get(char, 0) + 1
    return freq

print(char_frequency("Vidya Sagar Reddy"))
```

```python
def update_values(dictionary, updates):
    """
    Update dictionary values based on a list of (key, value) pairs.
    
    Args:
        dictionary (dict): The dictionary to update.
        updates (list of tuples): Each tuple contains (key, value).
    
    Returns:
        dict: The updated dictionary.
    """
    # Use dict.update with a dict comprehension for clarity and efficiency
    dictionary.update({key: value for key, value in updates})
    return dictionary

data = {'a': 1, 'b': 2, 'c': 3}
updates = [('b', 20), ('d', 40)]

print(update_values(data, updates))
```

```python
def calculate_total_value(inventory):
    """
    Calculate the total value of all items in the inventory
    based on their price and quantity.

    Args:
        inventory (dict): Nested dictionary of categories and items.

    Returns:
        int/float: Total value of all items.
    """
    # Use nested comprehensions to avoid explicit loops
    return sum(
        item["price"] * item["quantity"]
        for category in inventory.values()
        for item in category.values()
    )

inventory = {
    "fruits": {
        "apple": {"price": 2, "quantity": 10},
        "banana": {"price": 1, "quantity": 20},
    },
    "vegetables": {
        "carrot": {"price": 3, "quantity": 5},
        "potato": {"price": 2, "quantity": 15},
    },
}

print(calculate_total_value(inventory))
```
