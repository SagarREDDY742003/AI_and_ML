```python
def char_frequency(s):
    """
    Calculate the frequency of each character in a string.
    
    Args:
        s (str): Input string.
    
    Returns:
        dict: A dictionary mapping characters to their frequency.
    """
    freq = {}
    for char in s:
        # Use dict.get() to simplify increment logic
        freq[char] = freq.get(char, 0) + 1
    return freq

print(char_frequency("Vidya Sagar Reddy"))
```

```python
def update_dictionary(d, key, increment):
    """
    Update dictionary value for a given key by incrementing it.
    If the key does not exist, initialize it with the increment.
    
    Args:
        d (dict): Dictionary to update.
        key: Key to update.
        increment (int/float): Value to add.
    
    Returns:
        dict: Updated dictionary.
    """
    # Option 1: Using get()
    d[key] = d.get(key, 0) + increment
    
    # Option 2 (alternative): Using setdefault()
    # d[key] = d.setdefault(key, 0) + increment
    
    return d

data = {"a": 5, "b": 10}
print(update_dictionary(data, "a", 3))   # {'a': 8, 'b': 10}
print(update_dictionary(data, "c", 7))   # {'a': 8, 'b': 10, 'c': 7}
```

```python
import json

def convert_to_json(dict_list):
    """
    Convert a list of dictionaries to a JSON string.
    
    Args:
        dict_list (list): List of dictionaries.
    
    Returns:
        str: JSON string representation of the list.
    """
    # Validate input
    if not isinstance(dict_list, list) or not all(isinstance(d, dict) for d in dict_list):
        raise ValueError("Input must be a list of dictionaries")
    
    # Serialize the entire list at once
    return json.dumps(dict_list)

data = [
    {"name": "Alice", "age": 25},
    {"name": "Bob", "age": 30}
]

print(convert_to_json(data))
```

```python
from collections import defaultdict, OrderedDict

def process_transactions(transactions):
    """ 
    Process a list of (customer, amount) transactions and calculate 
    the total spent by each customer, preserving the order of first appearance. 
    
    Args: 
        transactions (list of tuples): Each tuple is (customer, amount). 
        
    Returns: 
        OrderedDict: Keys are customers in order of first transaction, 
                        values are their total amounts. 
    """
    customer_totals = OrderedDict()
    
    for customer, amount in transactions:
        customer_totals[customer] = customer_totals.get(customer,0) + amount
    return customer_totals

transactions = [('Alice', 50), ('Bob', 20), ('Alice', 30), ('Bob', 40), ('Charlie', 25)]
print(process_transactions(transactions))
```
