## Write a Python function that reads a list of filenames and attempts to open each file, using try-except blocks to handle any errors that occur. The function should return a dictionary with filenames as keys and either the file content or an error message as values.

````python
from pathlib import Path

def read_files(filenames):
    """
    Read multiple files and return a dictionary with filenames as keys
    and either their contents or an error message as values.

    Args:
        filenames (List[str]): List of file paths to read.

    Returns:
        Dict[str, str]: Dictionary mapping filenames to contents or error messages.
    """
    results = {}
    
    for file_name in filenames:
        path = Path(file_name)
        
        try:
            with path.open('r') as file:
                results[file_name] = file.read()
        except FileNotFoundError:
            results[file_name] = "File not found"
        except PermissionError:
            results[file_name] = "Permission denied"
        except Exception as e:
            results[file_name] = f"Error: {e}"

    return results

files = ["sample.txt", "sample1.txt", "post.csv"]
result = read_files(files)
print(result)
````

# Logging Errors

In Python, the **`logging` module** provides a flexible framework for tracking events that happen during program execution. It’s much more powerful and configurable than using `print()` statements for debugging, because it supports different severity levels, output destinations, and formatting.

---

## 🔑 Logging Basics
```python
import logging

# Configure logging
logging.basicConfig(
    level=logging.DEBUG,  # Minimum level to capture
    format="%(asctime)s - %(levelname)s - %(message)s"
)

# Example log messages
logging.debug("This is a debug message")       # Detailed info, for developers
logging.info("This is an info message")        # General information
logging.warning("This is a warning message")   # Something unexpected, but not critical
logging.error("This is an error message")      # Serious problem
logging.critical("This is a critical message") # Severe error, program may not continue
```

---

## 📂 Logging Levels
| Level       | Purpose                                                                 |
|-------------|-------------------------------------------------------------------------|
| `DEBUG`     | Detailed diagnostic information (lowest level).                         |
| `INFO`      | Confirmation that things are working as expected.                       |
| `WARNING`   | An indication of something unexpected, but the program can continue.    |
| `ERROR`     | A serious issue that prevents part of the program from functioning.     |
| `CRITICAL`  | A very serious error, program may not be able to continue running.      |

---

## ⚙️ Advanced Configuration
You can log to files instead of the console:

```python
logging.basicConfig(
    filename="app.log",          # Log file
    filemode="a",                # Append mode
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")
logging.error("Something went wrong")
```

---

## ✅ Example Use Case
Imagine processing numbers from a file:

```python
def process_numbers(numbers):

    for num in numbers:
        try:

            logging.error(f"Error processing {num}: {e}")
```

---

### 🎯 Why Use Logging Instead of `print()`?
- **Granularity**: Different severity levels.  
- **Flexibility**: Can log to console, files, or external systems.  
- **Control**: Easily enable/disable or filter logs without changing code.  
- **Professionalism**: Standard practice in production applications.  


### Design a Python function that reads a list of numbers from a file, processes each number, and logs any errors encountered during processing using the logging module. The function should handle file-related errors and number processing errors gracefully.
#### The function should read numbers from a file named 'numbers.txt'. Each line in the file contains a single number. The function should attempt to convert each line to an integer and log any ValueError encountered. Additionally, it should log any IOError if the file cannot be opened.

````python
import logging
from pathlib import Path

# Configure logging
logging.basicConfig(
    filename="processing.log",       # Log file
    level=logging.INFO,              # Log level
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def process_numbers_from_file(file_path = "numbers.txt"):
    """
    Read numbers from a file, process each number, and log errors.

    Args:
        file_path (str): Path to the file containing numbers (default 'numbers.txt').

    Returns:
        List[Union[int, str]]: List of processed results or error messages.
    """
    path = Path(file_path)
    results = []

    try:
        with path.open("r", encoding="utf-8") as file:
            for line_number, line in enumerate(file, start=1):
                try:
                    number = int(line.strip())
                    # Example processing: square the number
                    results.append(number ** 2)
                except ValueError:
                    error_msg = f"Line {line_number}: Invalid number format -> {line.strip()}"
                    logging.error(error_msg)
                    logging.warning(error_msg)
                    results.append(error_msg)
    except FileNotFoundError:
        error_msg = f"File not found: {file_path}"
        logging.error(error_msg)
        results.append(error_msg)
    except PermissionError:
        error_msg = f"Permission denied: {file_path}"
        logging.error(error_msg)
        results.append(error_msg)
    except IOError as e:
        error_msg = f"I/O error while reading {file_path}: {e}"
        logging.error(error_msg)
        results.append(error_msg)

    return results

print(process_numbers_from_file())
````

# Assertion Statements

In Python, **assertion statements** are a way to test assumptions in your code. They act like internal self‑checks: if the condition you assert is `True`, the program continues; if it’s `False`, Python raises an `AssertionError`.

---

### 🔑 Syntax
```python
assert condition, optional_message
```

- **condition** → an expression that should evaluate to `True`.
- **optional_message** → a string message displayed if the assertion fails.

---

### ✅ Example
```python
def divide(a, b):
    assert b != 0, "Denominator must not be zero"
    return a / b

print(divide(10, 2))   # Works fine
print(divide(10, 0))   # Raises AssertionError: Denominator must not be zero
```

---

### ⚡ Key Points
- Assertions are mainly used for **debugging and development**, not for handling runtime errors in production.
- They help catch **logical mistakes** early by enforcing assumptions.
- If Python is run with the `-O` (optimize) flag, **assert statements are ignored** — so they should not replace proper error handling.

---

### ✅ When to Use
- Checking preconditions (e.g., input must be positive).
- Validating invariants during algorithm execution.
- Ensuring postconditions after a function runs.

---

So, assertions are like saying: *“I believe this must be true here — if not, stop and tell me.”*  

#### Design a Python function that processes a list of operations, where each operation is a tuple containing an operator ('add', 'subtract', 'multiply', 'divide') and two operands. Implement robust error handling to manage invalid operations and division by zero. The function should return a list of results or error messages for each operation.

````python
def process_operations(operations):
    """
    Process a list of operations and return results or error messages.

    Args:
        operations (List[Tuple[str, float, float]]): 
            Each tuple contains (operator, operand1, operand2).
            Operator must be one of: 'add', 'subtract', 'multiply', 'divide'.

    Returns:
        List[Union[float, str]]: Results or error messages for each operation.
    """
    
    result = []
    
    for index, (operator, a, b) in enumerate(operations, start=1):
        try:
            if operator == "add":
                result.append(a + b)
            elif operator == "subtract":
                result.append(a - b)
            elif operator == "multiply":
                result.append(a * b)
            elif operator == "divide":
                if b==0:
                    raise ZeroDivisionError("Divison by zero")
                result.append(a/b)
            else:
                raise ValueError(f"Invalid operator '{operator}'")
        except ZeroDivisionError as e:
            result.append(f"Operation {index}: Error - {e}")
        except ValueError as e:
            result.append(f"Operation {index}: Error - {e}")
        except Exception as e:
            result.append(f"Operation {index}: Unexpected error - {e}")

    return result
        
ops = [
    ("add", 10, 5),
    ("subtract", 7, 2),
    ("multiply", 3, 4),
    ("divide", 8, 0),       # Division by zero
    ("power", 2, 3)         # Invalid operator
]

print(process_operations(ops))
````
